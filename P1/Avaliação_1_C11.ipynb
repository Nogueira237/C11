**Questão 1**

Mostre qual a média de idade das mulheres presentes neste dataset;


```python
import numpy as np
dataset = np.loadtxt('shopping_trends.csv', delimiter = ',', dtype = 'str')

# extrai coluna age e gender
dataset = dataset[1:,[1,2]]

cond = np.char.find(dataset[:,1], 'Female') != -1

mulher = dataset[cond]
idades = mulher[:,0].astype(int)
media = np.sum(idades)/len(idades)
print(f"A média de idade das mulheres presentes neste dataset é: {media:.2f}")

```

**Questão 2**
Mostre quantos clientes homens gastaram mais que a média de gastos das compras
deste dataset;


```python
import numpy as np
dataset = np.loadtxt('shopping_trends.csv', delimiter = ',', dtype = 'str')

dataset = dataset[1:, [2,5]]
gastos = dataset[:,1].astype(int)

media = np.sum(gastos)/len(gastos)
print(f"Média de gastos: {media:.2f}")

cond = np.char.find(dataset[:,0], 'Male') != -1

homem = dataset[cond]
gastosHomem = homem[:,1].astype(int)
maiorQueMedia =  gastosHomem > media

print(f"{np.sum(maiorQueMedia)} homens gastaram mais que a média.")
```

**Questão 3**

Mostre qual é a porcentagem de vendas do item menos vendido da loja

```python
import numpy as np
dataset = np.loadtxt('shopping_trends.csv', delimiter = ',', dtype = 'str')

dataset = dataset[1:, [15, 17]]

nVendas = dataset[:,0].astype(int)

indice = np.argmin(nVendas)

print(f"Número de vendas: {nVendas[indice]}")

totalVendas = np.sum(nVendas)
print(f"A porcentagem de venda desse item é: {100/totalVendas}%")
```

**Questão 4**
Mostre qual a porcentagem de vendas que tiveram algum tipo de desconto;

```python
import numpy as np
dataset = np.loadtxt('shopping_trends.csv', delimiter = ',', dtype = 'str')

dataset = dataset[1:, 13]

teveDesconto = np.char.find(dataset,'Yes') != -1

total = np.sum(teveDesconto)
media = total/len(teveDesconto) * 100
print(f"A porcentagem de vendas que tiveram desconto é: {media:.2f}%")
```

**Questão 5**
Mostre qual a cor de roupa mais popular no verão segundo este dataset.


```python
import numpy as np
dataset = np.loadtxt('shopping_trends.csv', delimiter = ',', dtype = 'str')

dataset = dataset[1:, [8,9]]

cond = np.char.find(dataset[:,1], 'Summer') != -1

summer = dataset[cond]

popular = np.unique(summer[:,0], return_counts=True)

indice = np.argmax(popular[1])

cores = popular[0]
print(f"A cor de roupa mais popular no verão é: {cores[indice]}")
```
